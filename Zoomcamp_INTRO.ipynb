```python
# 1. Select cars from Asia
asia = df[df["origin"] == "Asia"]

# 2 & 3. Select columns and first 7 rows
X = asia[["vehicle_weight", "model_year"]].head(7)

# 4. Convert to NumPy array
X = X.to_numpy()

# 5. Calculate XTX
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Calculate w
w = XTX_inv @ X.T @ y

# 9. Sum w
print("w =", w)
print("sum =", w.sum())
```
